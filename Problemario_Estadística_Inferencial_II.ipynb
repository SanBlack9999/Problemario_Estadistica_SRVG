# Unidad 1. Regresión lineal simple y correlación

## Medidas de dispersión
Suma de los cuadrados de las desviaciones de los valores de $X$ con respecto a su media:
$$ S_{xx} = \sum_{i=1}^{n} (X_i - \bar{X})^2 $$
Suma de los productos de las desviaciones de los valores de $X$ y $Y$ con respecto a sus medias:
$$S_{xy} = \sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})$$
Suma de los cuadrados de las desviaciones de los valores de $Y$ con respecto a su media:
$$ S_{yy} = \sum_{i=1}^{n} (Y_i - \bar{Y})^2 $$


## Coeficiente de correlación y de determinación
Coeficiente de correlación de Pearson:
$$ r = \frac{S_{xy}}{\sqrt{S_{xx} \cdot S_{yy}}} $$
Coeficiente de determinación:
$$ r^2 $$

## Recta de regresión ajustada
La regresión lineal ajustada se representa mediante estadísticos:
$$ \hat{Y} = b_0 + b_1 X $$
donde $\hat{Y}$ representa el valor de $Y$ obtenido mediante la recta de regresión ajustada (no la verdadera $Y$). Los estadísticos $b_0$ y $b_1$ se calculan de la siguiente manera:

$$b_1 = \frac{S_{xy}}{S_{xx}}$$

$$ b_0 = \bar{Y} - b_1 \bar{X} $$

## Cálculo de residuales
Residuales:
$$e_i = Y_i - \hat{Y}_i$$

## Sumas de cuadrados SS (Sum of Squares)
Suma de los Cuadrados de los Errores:
$$ SSE = \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2 = \sum_{i=1}^n e_i^2$$
Suma total de cuadrados:
$$SST = \sum_{i=1}^{n} (Y_i - \bar{Y})^2$$
Suma de cuadrados de regresión:
$$SSR = \sum_{i=1}^{n} (\hat{Y_i} - \bar{Y})^2$$
- SST: Mide la variabilidad total de los datos observados.
- SSR: Mide la variabilidad de los datos que el modelo de regresión explica.
- SSE: Mide la variabilidad no explicada por el modelo (es decir, los residuos).


## Intervalo de confianza
Estadístico de prueba _t_:
$$t = \frac{b_1}{SE(b_1)}$$
Error estándar de $b_1$:
$$SE(b_1) = \frac{\sqrt{SSE / (n-2)}}   {\sqrt{S_{xx}} }$$
o
$$SE(b_1) = \frac{\sqrt{\sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2  / (n-2)}}   {\sqrt{S_{xx}} }$$

Intervalo de confianza para $b_1$:
$$b_1 - t_{\alpha/2} \cdot SE(b_1) \lt \beta_1 \lt b_1 + t_{\alpha/2} \cdot SE(b_1)$$
donde $n$ representa la cantidad de pares de datos.

## Comprobación de supuestos
Comprobar suposiciones:
- Test de shapiro a los residuales $e_i$: Para comprobar si la distribución es normal sobre la recta.
- Grafico $X$ vs $Y$: Para observar si los datos soportan la suposición de linealidad.
- Gráfico de residuales: Para observar si los datos soportan la suposición de linealidad, complementario al coeficiente de correlación
- Test de Breusch-Pagan: Para detectar heteroscedasticidad en regresion lineal

Test de Shapiro:<br>
`from scipy.stats import shapiro`<br>
Después, se obtiene el valor-p:<br>
`_, valor_p_sh = shapiro(data)`<br>
- $H_0$: Los datos siguen una distribución normal
- $H_1$: Los datos no siguen una distribución normal

Test de Breusch-Pagan:<br>
`from statsmodels.stats.api import het_breuschpagan`<br>
Después, se obtiene el valor-p:<br>
`_, valor_p_bp, _, _ = het_breuschpagan(residuales, X)`
- $H_0$: Hay homoscedasticidad
- $H_1$: Hay heteroscedasticidad

## ANOVA en regresión lineal

| Fuente de variación | Suma de cuadrados (SS) | Grados de libertad (df) | Promedio de los cuadrados (MS) | Estadístico F |
|---------------------|------------------------|--------------------------|--------------------------------|---------------|
| Regresión  | $ SSR$ | $p$ | $MSR = \frac{SSR}{p}$ | $F = \frac{MSR}{MSE}$|
| Error | $SSE$| $n - p - 1 $| $MSE = \frac{SSE}{n - p - 1} $|     |
| Total | $SST $| $n - 1$|             |        |

donde $p$ es el número de parámetros para la recta de regresión ajustada (en la regresión simple $p=1$). Las hipótesis son:
$$H_0: \beta_1 = 0$$
$$H_1: \beta_1 \ne 0 $$

## Problemario de la Unidad 1

### Problema 1
Un profesor intenta mostrar a sus estudiantes la importancia de los exámenes cortos, aun cuando el 90% de la calificación final esté determinada por los exámenes parciales. Él cree que cuanto más altas sean las calificaciones de los exámenes cortos, más alta será la calificación final. Seleccionó una muestra aleatoria de 15 estudiantes de su clase con los siguientes datos:

| Promedio de exámenes cortos | Promedio final |
|--------------|--------------|
| 59 | 64 |
| 92 | 84 |
| 72 | 77 |
| 90 | 80 |
| 95 | 77 |
| 87 | 81 |
| 89 | 80 |
| 77 | 84 |
| 76 | 80 |
| 65 | 69 |
| 97 | 83 |
| 42 | 40 |
| 94 | 78 |
| 62 | 65 |
| 91 | 90 |



1.   Establesca una variable dependiente ($Y$) y una variable independiente ($X$).
2.   Realice un diagrama de dispersión para estos datos.
3. ¿Los datos soportan la suposición de linealidad?
4. Calcule el coeficiente de correlación e interprete el resultado.
5. Calcule el coeficiente de determinación e interprete el resultado.
6. Obtenga la recta de regresión ajustada y grafíquelo sobre el gráfico de dispersión.
7. Obtenga un intervalo de confianza del 95% para la pendiente de la recta de regresión ajustada ($b_1$)
8. Calcule los residuales y trace un nuevo gráfico de dispersión. Comente, ¿Parece que se verifican los supuestos?
9. Realice la prueba de Shapiro para los residuales y comente el resultado.
10. Realice la prueba de Brausch-Pagan para los residuales y comente el resultado.
11. Tres estudiantes sacaron 70, 75 y 84 de calificación. Según la recta de regresión ajustada, ¿cuáles son los resultados esperados para estos tres alumnos?
12. Realice una tabla ANOVA e interprete el resultado.



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress, shapiro, t, f
import statsmodels.api as sm
from statsmodels.stats.api import het_breuschpagan

# DATOS
X = np.array([59, 92, 72, 90, 95, 87, 89, 77, 76, 65, 97, 42, 94, 62, 91])
Y = np.array([64, 84, 77, 80, 77, 81, 80, 84, 80, 69, 83, 40, 78, 65, 90])
n = len(X)

# 1. VARIABLES
print("1. VARIABLES")
print("   X (independiente): Promedio de exámenes cortos")
print("   Y (dependiente):   Promedio final\n")

# 2. DIAGRAMA DE DISPERSIÓN
plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(X, Y, marker='o', color='steelblue', edgecolor='black', alpha=0.8, s=60)
plt.xlabel('Promedio de exámenes cortos (X)', fontsize=12)
plt.ylabel('Promedio final (Y)', fontsize=12)
plt.title('2. Diagrama de dispersión', fontsize=13)
plt.xticks(fontsize=10); plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

# 3. LINEALIDAD
print("3. LINEALIDAD")
print("   Los puntos muestran tendencia positiva aproximadamente lineal.")
print("   Los datos SÍ soportan el supuesto de linealidad.\n")

# 4. COEFICIENTE DE CORRELACIÓN
r, p_valor_r = pearsonr(X, Y)
print("4. COEFICIENTE DE CORRELACIÓN")
print(f"   r       = {r:.4f}")
print(f"   valor-p = {p_valor_r:.4f}")
print(f"   → Correlación positiva {'fuerte' if r >= 0.7 else 'moderada'}.\n")

# 5. COEFICIENTE DE DETERMINACIÓN
R2 = r**2
print("5. COEFICIENTE DE DETERMINACIÓN")
print(f"   R² = {R2:.4f}  ({R2*100:.2f}% de la variabilidad de Y explicada por X)\n")

# 6. RECTA DE REGRESIÓN
b1, b0, _, _, se_b1 = linregress(X, Y)
Y_estimado = b0 + b1 * X

print("6. RECTA DE REGRESIÓN AJUSTADA")
print(f"   b0 (intercepto) = {b0:.4f}")
print(f"   b1 (pendiente)  = {b1:.4f}")
print(f"   Ŷ = {b0:.4f} + {b1:.4f}·X\n")

X_linea = np.linspace(X.min() - 2, X.max() + 2, 100)
Y_linea = b0 + b1 * X_linea

plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(X, Y, marker='o', color='steelblue', edgecolor='black', alpha=0.8, s=60, label='Datos observados')
plt.plot(X_linea, Y_linea, color='crimson', linewidth=1.8, linestyle='--', label=f'Ŷ = {b0:.2f} + {b1:.2f}X')
plt.xlabel('Promedio de exámenes cortos (X)', fontsize=12)
plt.ylabel('Promedio final (Y)', fontsize=12)
plt.title('6. Recta de regresión ajustada', fontsize=13)
plt.legend(fontsize=10, loc='best', frameon=True)
plt.xticks(fontsize=10); plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

# 7. INTERVALO DE CONFIANZA AL 95% PARA b1
gl = n - 2
t_critico = t.ppf(0.975, df=gl)
ic_inf = b1 - t_critico * se_b1
ic_sup = b1 + t_critico * se_b1

print("7. INTERVALO DE CONFIANZA AL 95% PARA b1")
print(f"   b1          = {b1:.4f}")
print(f"   Error est.  = {se_b1:.4f}")
print(f"   t crítico   = {t_critico:.4f}  (gl={gl})")
print(f"   IC 95%: [{ic_inf:.4f},  {ic_sup:.4f}]\n")

# 8. RESIDUALES
residuales = Y - Y_estimado

print("8. RESIDUALES")
print(f"   {'Obs':>4} {'X':>5} {'Y':>5} {'Ŷ':>8} {'e':>8}")
print("   " + "-"*34)
for i in range(n):
    print(f"   {i+1:>4} {X[i]:>5} {Y[i]:>5} {Y_estimado[i]:>8.2f} {residuales[i]:>+8.2f}")
print()

plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(Y_estimado, residuales, marker='o', color='darkorange', edgecolor='black', alpha=0.8, s=60)
plt.axhline(y=0, color='black', linestyle='--', linewidth=1.0)
plt.xlabel('Valores ajustados (Ŷ)', fontsize=12)
plt.ylabel('Residuales (e)', fontsize=12)
plt.title('8. Residuales vs. Valores ajustados', fontsize=13)
plt.xticks(fontsize=10); plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

# 9. PRUEBA DE SHAPIRO-WILK
stat_sw, p_valor_sw = shapiro(residuales)
print("9. PRUEBA DE SHAPIRO-WILK (normalidad)")
print(f"   W       = {stat_sw:.4f}")
print(f"   valor-p = {p_valor_sw:.4f}")
if p_valor_sw > 0.05:
    print("   → No se rechaza H0: se asume NORMALIDAD en los residuales.\n")
else:
    print("   → Se rechaza H0: los residuales NO son normales.\n")

# 10. PRUEBA DE BREUSCH-PAGAN
X_constante = sm.add_constant(X)
_, p_valor_bp, _, _ = het_breuschpagan(residuales, X_constante)
print("10. PRUEBA DE BREUSCH-PAGAN (homoscedasticidad)")
print(f"    valor-p = {p_valor_bp:.4f}")
if p_valor_bp > 0.05:
    print("    → No se rechaza H0: se asume HOMOSCEDASTICIDAD.\n")
else:
    print("    → Se rechaza H0: existe HETEROCEDASTICIDAD.\n")

# 11. PREDICCIONES
X_nuevos = np.array([70, 75, 84])
Y_pred   = b0 + b1 * X_nuevos
print("11. PREDICCIONES")
for xi, yi in zip(X_nuevos, Y_pred):
    print(f"    X = {xi}  →  Ŷ = {b0:.2f} + {b1:.2f}·{xi} = {yi:.2f}")
print()

# 12. TABLA ANOVA
Y_media = np.mean(Y)
SST = np.sum((Y - Y_media)**2)
SSR = np.sum((Y_estimado - Y_media)**2)
SSE = np.sum((Y - Y_estimado)**2)
gl_R, gl_E, gl_T = 1, n - 2, n - 1
MSR = SSR / gl_R
MSE = SSE / gl_E
F_stat    = MSR / MSE
p_valor_F = 1 - f.cdf(F_stat, dfn=gl_R, dfd=gl_E)

print("12. TABLA ANOVA")
print("=" * 65)
print(f"{'Fuente':<18} {'SC':>10} {'gl':>5} {'CM':>12} {'F':>8} {'p-valor':>9}")
print("=" * 65)
print(f"{'Regresión':<18} {SSR:>10.4f} {gl_R:>5} {MSR:>12.4f} {F_stat:>8.4f} {p_valor_F:>9.4f}")
print(f"{'Error':<18} {SSE:>10.4f} {gl_E:>5} {MSE:>12.4f}")
print("-" * 65)
print(f"{'Total':<18} {SST:>10.4f} {gl_T:>5}")
print("=" * 65)
if p_valor_F < 0.05:
    print(f"\n→ Se rechaza H0 (p = {p_valor_F:.4f} < 0.05): el modelo es SIGNIFICATIVO.")
else:
    print(f"\n→ No se rechaza H0 (p = {p_valor_F:.4f} ≥ 0.05): modelo no SIFNIFICATIVO.")
```

### Problema 2
William Hawkins, vicepresidente de personal de la International Motors, trabaja en la relación entre el salario de un trabajador y el porcentaje de ausentismo. Hawkins dividió el intervalo de salarios de International en 12 grados o niveles (1 es el menor grado, 12 el más alto) y después muestreó aleatoriamente a un grupo de trabajadores. Determinó el grado de salario de cada trabajador y el número de días que ese empleado había faltado en los últimos 3 años.

| Categoría de salario | 11 | 10 | 8  | 5  | 9  | 7  | 3  |
|----------------------|----|----|----|----|----|----|----|
| Ausencias           | 18 | 17 | 29 | 36 | 11 | 28 | 35 |

| Categoría de salario | 11 | 8  | 7  | 2  | 9  | 8  | 3  |
|----------------------|----|----|----|----|----|----|----|
| Ausencias           | 14 | 20 | 32 | 39 | 16 | 31 | 40 |

1.   Establesca una variable dependiente ($Y$) y una variable independiente ($X$).
2.   Realice un diagrama de dispersión para estos datos.
3. ¿Los datos soportan la suposición de linealidad?
4. Calcule el coeficiente de correlación e interprete el resultado.
5. Calcule el coeficiente de determinación e interprete el resultado.
6. Obtenga la recta de regresión ajustada y grafíquelo sobre el gráfico de dispersión.
7. Obtenga un intervalo de confianza del 95% para la pendiente de la recta de regresión ajustada ($b_1$)
8. Calcule los residuales y trace un nuevo gráfico de dispersión. Comente, ¿Parece que se verifican los supuestos?
9. Realice la prueba de Shapiro para los residuales y comente el resultado.
10. Realice la prueba de Brausch-Pagan para los residuales y comente el resultado.
11. Utiliza la recta de regresión para interpolar dos valores y extrapolar uno. Comenta estos resultados.
12. Realice una tabla ANOVA e interprete el resultado.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress, shapiro, t, f
import statsmodels.api as sm
from statsmodels.stats.api import het_breuschpagan

#DATOS
X = np.array([11, 10, 8, 5, 9, 7, 3, 11, 8, 7, 2, 9, 8, 3])
Y = np.array([18, 17, 29, 36, 11, 28, 35, 14, 20, 32, 39, 16, 31, 40])
n = len(X)

# 1. VARIABLES
print("1. VARIABLES")
print("   X (independiente): Categoría de salario (1 = menor, 12 = mayor)")
print("   Y (dependiente):   Número de días de ausencia en 3 años")
print("   Hipótesis: a mayor salario, menor ausentismo (relación negativa)\n")

# 2. DIAGRAMA DE DISPERSIÓN
plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(X, Y, marker='o', color='steelblue', edgecolor='black', alpha=0.8, s=60)
plt.xlabel('Categoría de salario (X)', fontsize=12)
plt.ylabel('Días de ausencia (Y)', fontsize=12)
plt.title('2. Diagrama de dispersión', fontsize=13)
plt.xticks(range(1, 13), fontsize=10)
plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

# 3. LINEALIDAD
print("3. LINEALIDAD")
print("   El diagrama muestra una tendencia negativa aproximadamente lineal.")
print("   Los datos SÍ soportan el supuesto de linealidad.\n")

# 4. COEFICIENTE DE CORRELACIÓN
r, p_valor_r = pearsonr(X, Y)
print("4. COEFICIENTE DE CORRELACIÓN")
print(f"   r       = {r:.4f}")
print(f"   valor-p = {p_valor_r:.4f}")
if abs(r) >= 0.7:
    fuerza = "fuerte"
elif abs(r) >= 0.4:
    fuerza = "moderada"
else:
    fuerza = "débil"
direccion = "negativa" if r < 0 else "positiva"
print(f"   → Correlación lineal {direccion} {fuerza}.")
print(f"     Los trabajadores con mayor categoría salarial tienden a ausentarse menos.\n")

# 5. COEFICIENTE DE DETERMINACIÓN
R2 = r**2
print("5. COEFICIENTE DE DETERMINACIÓN")
print(f"   R² = {R2:.4f}  →  el {R2*100:.2f}% de la variabilidad en los días de")
print(f"   ausencia es explicada por la categoría de salario.\n")

# 6. RECTA DE REGRESIÓN
b1, b0, _, _, se_b1 = linregress(X, Y)
Y_estimado = b0 + b1 * X

print("6. RECTA DE REGRESIÓN AJUSTADA")
print(f"   b0 (intercepto) = {b0:.4f}")
print(f"   b1 (pendiente)  = {b1:.4f}")
print(f"   Ŷ = {b0:.4f} + ({b1:.4f})·X\n")

X_linea = np.linspace(X.min() - 0.5, X.max() + 0.5, 200)
Y_linea = b0 + b1 * X_linea

plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(X, Y, marker='o', color='steelblue', edgecolor='black',
            alpha=0.8, s=60, label='Datos observados')
plt.plot(X_linea, Y_linea, color='crimson', linewidth=1.8, linestyle='--',
         label=f'Ŷ = {b0:.2f} + ({b1:.2f})X')
plt.xlabel('Categoría de salario (X)', fontsize=12)
plt.ylabel('Días de ausencia (Y)', fontsize=12)
plt.title('6. Recta de regresión ajustada', fontsize=13)
plt.legend(fontsize=10, loc='best', frameon=True)
plt.xticks(range(1, 13), fontsize=10)
plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

# 7. INTERVALO DE CONFIANZA AL 95% PARA b1
gl = n - 2
t_critico = t.ppf(0.975, df=gl)
ic_inf = b1 - t_critico * se_b1
ic_sup = b1 + t_critico * se_b1

print("7. INTERVALO DE CONFIANZA AL 95% PARA b1")
print(f"   b1          = {b1:.4f}")
print(f"   Error est.  = {se_b1:.4f}")
print(f"   t crítico   = {t_critico:.4f}  (gl = {gl})")
print(f"   IC 95%: [{ic_inf:.4f},  {ic_sup:.4f}]")
print(f"   → El IC no contiene el 0: b1 es significativamente distinto de cero.\n")

# 8. RESIDUALES
residuales = Y - Y_estimado

print("8. RESIDUALES")
print(f"   {'Obs':>4} {'X':>5} {'Y':>5} {'Ŷ':>8} {'e':>8}")
print("   " + "-"*34)
for i in range(n):
    print(f"   {i+1:>4} {X[i]:>5} {Y[i]:>5} {Y_estimado[i]:>8.2f} {residuales[i]:>+8.2f}")
print()

plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(Y_estimado, residuales, marker='o', color='darkorange',
            edgecolor='black', alpha=0.8, s=60)
plt.axhline(y=0, color='black', linestyle='--', linewidth=1.0)
plt.xlabel('Valores ajustados (Ŷ)', fontsize=12)
plt.ylabel('Residuales (e)', fontsize=12)
plt.title('8. Residuales vs. Valores ajustados', fontsize=13)
plt.xticks(fontsize=10); plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

print("   Comentario: los residuales se distribuyen aleatoriamente alrededor de")
print("   cero sin patrones evidentes, lo que sugiere que los supuestos se verifican.\n")

# 9. PRUEBA DE SHAPIRO-WILK
stat_sw, p_valor_sw = shapiro(residuales)
print("9. PRUEBA DE SHAPIRO-WILK (normalidad de residuales)")
print(f"   W       = {stat_sw:.4f}")
print(f"   valor-p = {p_valor_sw:.4f}")
if p_valor_sw > 0.05:
    print("   → No se rechaza H0 (p > 0.05): se asume NORMALIDAD en los residuales.")
    print("     Con n = 14 ≤ 30, Shapiro-Wilk es la prueba más adecuada.\n")
else:
    print("   → Se rechaza H0 (p ≤ 0.05): los residuales NO son normales.\n")

# 10. PRUEBA DE BREUSCH-PAGAN
X_constante = sm.add_constant(X)
_, p_valor_bp, _, _ = het_breuschpagan(residuales, X_constante)
print("10. PRUEBA DE BREUSCH-PAGAN (homoscedasticidad)")
print(f"    valor-p = {p_valor_bp:.4f}")
if p_valor_bp > 0.05:
    print("    → No se rechaza H0 (p > 0.05): se asume HOMOSCEDASTICIDAD.\n")
else:
    print("    → Se rechaza H0 (p ≤ 0.05): existe HETEROCEDASTICIDAD.\n")

# 11. INTERPOLACIÓN Y EXTRAPOLACIÓN
X_min, X_max = X.min(), X.max()
casos = [(4, "Interpolación"), (6, "Interpolación"), (13, "Extrapolación")]

print("11. INTERPOLACIÓN Y EXTRAPOLACIÓN")
print(f"    Rango observado de X: [{X_min}, {X_max}]")
print(f"    Recta: Ŷ = {b0:.4f} + ({b1:.4f})·X\n")
print(f"    {'Tipo':<16} {'X':>4} {'Ŷ':>8}  Comentario")
print("    " + "-"*60)
for xi, tipo in casos:
    yi = b0 + b1 * xi
    comentario = "Predicción confiable (dentro del rango)" if tipo == "Interpolación" else "⚠ Usar con cautela (fuera del rango observado)"
    print(f"    {tipo:<16} {xi:>4} {yi:>8.2f}  {comentario}")
print()
print("    → La interpolación es confiable dentro del rango estudiado.")
print("      La extrapolación (X=13) excede los datos; usar con reserva.\n")

# 12. TABLA ANOVA
Y_media = np.mean(Y)
SST = np.sum((Y - Y_media)**2)
SSR = np.sum((Y_estimado - Y_media)**2)
SSE = np.sum((Y - Y_estimado)**2)
gl_R, gl_E, gl_T = 1, n - 2, n - 1
MSR = SSR / gl_R
MSE = SSE / gl_E
F_stat    = MSR / MSE
p_valor_F = 1 - f.cdf(F_stat, dfn=gl_R, dfd=gl_E)

print("12. TABLA ANOVA")
print("=" * 65)
print(f"{'Fuente':<18} {'SC':>10} {'gl':>5} {'CM':>12} {'F':>8} {'p-valor':>9}")
print("=" * 65)
print(f"{'Regresión':<18} {SSR:>10.4f} {gl_R:>5} {MSR:>12.4f} {F_stat:>8.4f} {p_valor_F:>9.4f}")
print(f"{'Error':<18} {SSE:>10.4f} {gl_E:>5} {MSE:>12.4f}")
print("-" * 65)
print(f"{'Total':<18} {SST:>10.4f} {gl_T:>5}")
print("=" * 65)
if p_valor_F < 0.05:
    print(f"\n→ Se rechaza H0 (p = {p_valor_F:.4f} < 0.05): el modelo es SIGNIFICATIVO.")
    print("  La categoría de salario explica significativamente el ausentismo.")
else:
    print(f"\n→ No se rechaza H0 (p = {p_valor_F:.4f} ≥ 0.05): modelo no significativo.")
```

1.   Establesca una variable dependiente ($Y$) y una variable independiente ($X$).
2.   Realice un diagrama de dispersión para estos datos.

5. Calcule el coeficiente de determinación e interprete el resultado.

7. Obtenga un intervalo de confianza del 95% para la pendiente de la recta de regresión ajustada ($b_1$)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress, shapiro, t, f
import statsmodels.api as sm
from statsmodels.stats.api import het_breuschpagan

# DATOS
X = np.array([40, 42, 53, 35, 56, 39, 48, 30, 37, 40])
Y = np.array([191, 170, 272, 155, 280, 173, 234, 116, 153, 178])
n = len(X)

# 1. VARIABLES
print("1. VARIABLES")
print("   X (independiente): Unidades producidas")
print("   Y (dependiente):   Gastos generales")
print("   Hipótesis: a mayor producción, mayores gastos generales (relación positiva)\n")

# 2. DIAGRAMA DE DISPERSIÓN
plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(X, Y, marker='o', color='steelblue', edgecolor='black', alpha=0.8, s=60)
plt.xlabel('Unidades producidas (X)', fontsize=12)
plt.ylabel('Gastos generales (Y)', fontsize=12)
plt.title('2. Diagrama de dispersión', fontsize=13)
plt.xticks(fontsize=10); plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

# 3. LINEALIDAD
print("3. LINEALIDAD")
print("   El diagrama muestra una tendencia positiva aproximadamente lineal.")
print("   Los datos SÍ soportan el supuesto de linealidad.\n")

# 4. COEFICIENTE DE CORRELACIÓN
r, p_valor_r = pearsonr(X, Y)
print("4. COEFICIENTE DE CORRELACIÓN")
print(f"   r       = {r:.4f}")
print(f"   valor-p = {p_valor_r:.4f}")
if abs(r) >= 0.7:
    fuerza = "fuerte"
elif abs(r) >= 0.4:
    fuerza = "moderada"
else:
    fuerza = "débil"
direccion = "negativa" if r < 0 else "positiva"
print(f"   → Correlación lineal {direccion} {fuerza}.")
print(f"     Las plantas con mayor producción tienden a tener mayores gastos generales.\n")

# 5. COEFICIENTE DE DETERMINACIÓN
R2 = r**2
print("5. COEFICIENTE DE DETERMINACIÓN")
print(f"   R² = {R2:.4f}  →  el {R2*100:.2f}% de la variabilidad en los gastos generales")
print(f"   es explicada por el número de unidades producidas.\n")

# 6. RECTA DE REGRESIÓN
b1, b0, _, _, se_b1 = linregress(X, Y)
Y_estimado = b0 + b1 * X

print("6. RECTA DE REGRESIÓN AJUSTADA")
print(f"   b0 (intercepto) = {b0:.4f}")
print(f"   b1 (pendiente)  = {b1:.4f}")
print(f"   Ŷ = {b0:.4f} + {b1:.4f}·X\n")

X_linea = np.linspace(X.min() - 1, X.max() + 1, 200)
Y_linea = b0 + b1 * X_linea

plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(X, Y, marker='o', color='steelblue', edgecolor='black',
            alpha=0.8, s=60, label='Datos observados')
plt.plot(X_linea, Y_linea, color='crimson', linewidth=1.8, linestyle='--',
         label=f'Ŷ = {b0:.2f} + {b1:.2f}X')
plt.xlabel('Unidades producidas (X)', fontsize=12)
plt.ylabel('Gastos generales (Y)', fontsize=12)
plt.title('6. Recta de regresión ajustada', fontsize=13)
plt.legend(fontsize=10, loc='best', frameon=True)
plt.xticks(fontsize=10); plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

# 7. INTERVALO DE CONFIANZA AL 95% PARA b1
gl = n - 2
t_critico = t.ppf(0.975, df=gl)
ic_inf = b1 - t_critico * se_b1
ic_sup = b1 + t_critico * se_b1

print("7. INTERVALO DE CONFIANZA AL 95% PARA b1")
print(f"   b1          = {b1:.4f}")
print(f"   Error est.  = {se_b1:.4f}")
print(f"   t crítico   = {t_critico:.4f}  (gl = {gl})")
print(f"   IC 95%: [{ic_inf:.4f},  {ic_sup:.4f}]")
print(f"   → El IC no contiene el 0: b1 es significativamente distinto de cero.\n")

# 8. RESIDUALES
residuales = Y - Y_estimado

print("8. RESIDUALES")
print(f"   {'Obs':>4} {'X':>6} {'Y':>6} {'Ŷ':>9} {'e':>9}")
print("   " + "-"*38)
for i in range(n):
    print(f"   {i+1:>4} {X[i]:>6} {Y[i]:>6} {Y_estimado[i]:>9.2f} {residuales[i]:>+9.2f}")
print()

plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(Y_estimado, residuales, marker='o', color='darkorange',
            edgecolor='black', alpha=0.8, s=60)
plt.axhline(y=0, color='black', linestyle='--', linewidth=1.0)
plt.xlabel('Valores ajustados (Ŷ)', fontsize=12)
plt.ylabel('Residuales (e)', fontsize=12)
plt.title('8. Residuales vs. Valores ajustados', fontsize=13)
plt.xticks(fontsize=10); plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

print("   Comentario: los residuales se distribuyen aleatoriamente alrededor de")
print("   cero sin patrones sistemáticos, lo que sugiere que los supuestos se verifican.\n")

# 9. PRUEBA DE SHAPIRO-WILK
stat_sw, p_valor_sw = shapiro(residuales)
print("9. PRUEBA DE SHAPIRO-WILK (normalidad de residuales)")
print(f"   W       = {stat_sw:.4f}")
print(f"   valor-p = {p_valor_sw:.4f}")
if p_valor_sw > 0.05:
    print("   → No se rechaza H0 (p > 0.05): se asume NORMALIDAD en los residuales.")
    print("     Con n = 10 ≤ 30, Shapiro-Wilk es la prueba más adecuada.\n")
else:
    print("   → Se rechaza H0 (p ≤ 0.05): los residuales NO son normales.\n")

# 10. PRUEBA DE BREUSCH-PAGAN
X_constante = sm.add_constant(X)
_, p_valor_bp, _, _ = het_breuschpagan(residuales, X_constante)
print("10. PRUEBA DE BREUSCH-PAGAN (homoscedasticidad)")
print(f"    valor-p = {p_valor_bp:.4f}")
if p_valor_bp > 0.05:
    print("    → No se rechaza H0 (p > 0.05): se asume HOMOSCEDASTICIDAD.\n")
else:
    print("    → Se rechaza H0 (p ≤ 0.05): existe HETEROCEDASTICIDAD.\n")

# 11. INTERPOLACIÓN Y EXTRAPOLACIÓN
X_min, X_max = X.min(), X.max()
casos = [(38, "Interpolación"), (45, "Interpolación"), (65, "Extrapolación")]

print("11. INTERPOLACIÓN Y EXTRAPOLACIÓN")
print(f"    Rango observado de X: [{X_min}, {X_max}]")
print(f"    Recta: Ŷ = {b0:.4f} + {b1:.4f}·X\n")
print(f"    {'Tipo':<16} {'X':>4} {'Ŷ':>9}  Comentario")
print("    " + "-"*62)
for xi, tipo in casos:
    yi = b0 + b1 * xi
    comentario = "Predicción confiable (dentro del rango)" if tipo == "Interpolación" else "⚠ Usar con cautela (fuera del rango observado)"
    print(f"    {tipo:<16} {xi:>4} {yi:>9.2f}  {comentario}")
print()
print("    → Los valores interpolados (X=38 y X=45) están dentro del rango [30, 56].")
print("      El valor extrapolado (X=65) está fuera del rango; usar con reserva.\n")

# 12. TABLA ANOVA
Y_media = np.mean(Y)
SST = np.sum((Y - Y_media)**2)
SSR = np.sum((Y_estimado - Y_media)**2)
SSE = np.sum((Y - Y_estimado)**2)
gl_R, gl_E, gl_T = 1, n - 2, n - 1
MSR = SSR / gl_R
MSE = SSE / gl_E
F_stat    = MSR / MSE
p_valor_F = 1 - f.cdf(F_stat, dfn=gl_R, dfd=gl_E)

print("12. TABLA ANOVA")
print("=" * 65)
print(f"{'Fuente':<18} {'SC':>10} {'gl':>5} {'CM':>12} {'F':>8} {'p-valor':>9}")
print("=" * 65)
print(f"{'Regresión':<18} {SSR:>10.4f} {gl_R:>5} {MSR:>12.4f} {F_stat:>8.4f} {p_valor_F:>9.4f}")
print(f"{'Error':<18} {SSE:>10.4f} {gl_E:>5} {MSE:>12.4f}")
print("-" * 65)
print(f"{'Total':<18} {SST:>10.4f} {gl_T:>5}")
print("=" * 65)
if p_valor_F < 0.05:
    print(f"\n→ Se rechaza H0 (p = {p_valor_F:.4f} < 0.05): el modelo es SIGNIFICATIVO.")
    print("  Las unidades producidas explican significativamente los gastos generales.")
else:
    print(f"\n→ No se rechaza H0 (p = {p_valor_F:.4f} ≥ 0.05): modelo no significativo.")
```

### Problema 4
Las ventas de línea blanca varían según el estado del mercado de casas nuevas: cuando las ventas de casas nuevas son buenas, también lo son las de lavaplatos, lavadoras de ropa, secadoras y refrigeradores.
Una asociación de comercio compiló los siguientes datos históricos (en miles de unidades) de las ventas de línea blanca y la construcción de casas.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress, shapiro, t, f
import statsmodels.api as sm
from statsmodels.stats.api import het_breuschpagan

#  DATOS
X = np.array([2.0, 2.5, 3.2, 3.6, 3.7, 4.0, 4.2, 4.6, 4.8, 5.0])
Y = np.array([5.0, 5.5, 6.0, 7.0, 7.2, 7.7, 8.4, 9.0, 9.7, 10.0])
n = len(X)

# 1. VARIABLES
print("1. VARIABLES")
print("   X (independiente): Construcción de casas (miles de unidades)")
print("   Y (dependiente):   Ventas de línea blanca (miles de unidades)")
print("   Hipótesis: a mayor construcción de casas, mayores ventas de línea blanca.\n")

# 2. DIAGRAMA DE DISPERSIÓN
plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(X, Y, marker='o', color='steelblue', edgecolor='black', alpha=0.8, s=60)
plt.xlabel('Construcción de casas (miles)', fontsize=12)
plt.ylabel('Ventas de línea blanca (miles)', fontsize=12)
plt.title('2. Diagrama de dispersión', fontsize=13)
plt.xticks(fontsize=10); plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

# 3. LINEALIDAD
print("3. LINEALIDAD")
print("   El diagrama muestra una tendencia positiva muy clara y lineal.")
print("   Los datos SÍ soportan fuertemente el supuesto de linealidad.\n")

# 4. COEFICIENTE DE CORRELACIÓN
r, p_valor_r = pearsonr(X, Y)
print("4. COEFICIENTE DE CORRELACIÓN")
print(f"   r       = {r:.4f}")
print(f"   valor-p = {p_valor_r:.4f}")
if abs(r) >= 0.7:
    fuerza = "fuerte"
elif abs(r) >= 0.4:
    fuerza = "moderada"
else:
    fuerza = "débil"
direccion = "negativa" if r < 0 else "positiva"
print(f"   → Correlación lineal {direccion} {fuerza}.")
print(f"     Existe una relación lineal muy estrecha entre construcción y ventas.\n")

# 5. COEFICIENTE DE DETERMINACIÓN
R2 = r**2
print("5. COEFICIENTE DE DETERMINACIÓN")
print(f"   R² = {R2:.4f}  →  el {R2*100:.2f}% de la variabilidad en las ventas de")
print(f"   línea blanca es explicada por la construcción de casas nuevas.\n")

# 6. RECTA DE REGRESIÓN
b1, b0, _, _, se_b1 = linregress(X, Y)
Y_estimado = b0 + b1 * X

print("6. RECTA DE REGRESIÓN AJUSTADA")
print(f"   b0 (intercepto) = {b0:.4f}")
print(f"   b1 (pendiente)  = {b1:.4f}")
print(f"   Ŷ = {b0:.4f} + {b1:.4f}·X\n")

X_linea = np.linspace(X.min() - 0.2, X.max() + 0.2, 200)
Y_linea = b0 + b1 * X_linea

plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(X, Y, marker='o', color='steelblue', edgecolor='black',
            alpha=0.8, s=60, label='Datos observados')
plt.plot(X_linea, Y_linea, color='crimson', linewidth=1.8, linestyle='--',
         label=f'Ŷ = {b0:.2f} + {b1:.2f}X')
plt.xlabel('Construcción de casas (miles)', fontsize=12)
plt.ylabel('Ventas de línea blanca (miles)', fontsize=12)
plt.title('6. Recta de regresión ajustada', fontsize=13)
plt.legend(fontsize=10, loc='best', frameon=True)
plt.xticks(fontsize=10); plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

# 7. INTERVALO DE CONFIANZA AL 95% PARA b1
gl = n - 2
t_critico = t.ppf(0.975, df=gl)
ic_inf = b1 - t_critico * se_b1
ic_sup = b1 + t_critico * se_b1

print("7. INTERVALO DE CONFIANZA AL 95% PARA b1")
print(f"   b1          = {b1:.4f}")
print(f"   Error est.  = {se_b1:.4f}")
print(f"   t crítico   = {t_critico:.4f}  (gl = {gl})")
print(f"   IC 95%: [{ic_inf:.4f},  {ic_sup:.4f}]")
print(f"   → El IC no contiene el 0: b1 es significativamente distinto de cero.\n")

# 8. RESIDUALES
residuales = Y - Y_estimado

print("8. RESIDUALES")
print(f"   {'Obs':>4} {'X':>6} {'Y':>6} {'Ŷ':>9} {'e':>9}")
print("   " + "-"*38)
for i in range(n):
    print(f"   {i+1:>4} {X[i]:>6.1f} {Y[i]:>6.1f} {Y_estimado[i]:>9.4f} {residuales[i]:>+9.4f}")
print()

plt.figure(figsize=(6, 4), dpi=120)
plt.scatter(Y_estimado, residuales, marker='o', color='darkorange',
            edgecolor='black', alpha=0.8, s=60)
plt.axhline(y=0, color='black', linestyle='--', linewidth=1.0)
plt.xlabel('Valores ajustados (Ŷ)', fontsize=12)
plt.ylabel('Residuales (e)', fontsize=12)
plt.title('8. Residuales vs. Valores ajustados', fontsize=13)
plt.xticks(fontsize=10); plt.yticks(fontsize=10)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.grid(visible=True, linestyle='--', linewidth=0.7, alpha=0.5, color='gray')
plt.tight_layout(); plt.show()

print("   Comentario: los residuales se distribuyen aleatoriamente alrededor de")
print("   cero sin patrones de curvatura ni de embudo; los supuestos se verifican.\n")

# 9. PRUEBA DE SHAPIRO-WILK
stat_sw, p_valor_sw = shapiro(residuales)
print("9. PRUEBA DE SHAPIRO-WILK (normalidad de residuales)")
print(f"   W       = {stat_sw:.4f}")
print(f"   valor-p = {p_valor_sw:.4f}")
if p_valor_sw > 0.05:
    print("   → No se rechaza H0 (p > 0.05): se asume NORMALIDAD en los residuales.")
    print("     Con n = 10 ≤ 30, Shapiro-Wilk es la prueba más adecuada.\n")
else:
    print("   → Se rechaza H0 (p ≤ 0.05): los residuales NO son normales.\n")

# 10. PRUEBA DE BREUSCH-PAGAN
X_constante = sm.add_constant(X)
_, p_valor_bp, _, _ = het_breuschpagan(residuales, X_constante)
print("10. PRUEBA DE BREUSCH-PAGAN (homoscedasticidad)")
print(f"    valor-p = {p_valor_bp:.4f}")
if p_valor_bp > 0.05:
    print("    → No se rechaza H0 (p > 0.05): se asume HOMOSCEDASTICIDAD.\n")
else:
    print("    → Se rechaza H0 (p ≤ 0.05): existe HETEROCEDASTICIDAD.\n")

# 11. INTERPOLACIÓN Y EXTRAPOLACIÓN
X_min, X_max = X.min(), X.max()
casos = [(3.0, "Interpolación"), (4.4, "Interpolación"), (6.5, "Extrapolación")]

print("11. INTERPOLACIÓN Y EXTRAPOLACIÓN")
print(f"    Rango observado de X: [{X_min}, {X_max}]")
print(f"    Recta: Ŷ = {b0:.4f} + {b1:.4f}·X\n")
print(f"    {'Tipo':<16} {'X':>5} {'Ŷ':>9}  Comentario")
print("    " + "-"*64)
for xi, tipo in casos:
    yi = b0 + b1 * xi
    comentario = "Predicción confiable (dentro del rango)" if tipo == "Interpolación" else "⚠ Usar con cautela (fuera del rango observado)"
    print(f"    {tipo:<16} {xi:>5.1f} {yi:>9.4f}  {comentario}")
print()
print("    → Las interpolaciones (X=3.0 y X=4.4) están dentro del rango [2.0, 5.0].")
print("      La extrapolación (X=6.5) supera el máximo; usar con reserva.\n")

# 12. TABLA ANOVA
Y_media = np.mean(Y)
SST = np.sum((Y - Y_media)**2)
SSR = np.sum((Y_estimado - Y_media)**2)
SSE = np.sum((Y - Y_estimado)**2)
gl_R, gl_E, gl_T = 1, n - 2, n - 1
MSR = SSR / gl_R
MSE = SSE / gl_E
F_stat    = MSR / MSE
p_valor_F = 1 - f.cdf(F_stat, dfn=gl_R, dfd=gl_E)

print("12. TABLA ANOVA")
print("=" * 65)
print(f"{'Fuente':<18} {'SC':>10} {'gl':>5} {'CM':>12} {'F':>8} {'p-valor':>9}")
print("=" * 65)
print(f"{'Regresión':<18} {SSR:>10.4f} {gl_R:>5} {MSR:>12.4f} {F_stat:>8.4f} {p_valor_F:>9.4f}")
print(f"{'Error':<18} {SSE:>10.4f} {gl_E:>5} {MSE:>12.4f}")
print("-" * 65)
print(f"{'Total':<18} {SST:>10.4f} {gl_T:>5}")
print("=" * 65)
if p_valor_F < 0.05:
    print(f"\n→ Se rechaza H0 (p = {p_valor_F:.4f} < 0.05): el modelo es SIGNIFICATIVO.")
    print("  La construcción de casas explica significativamente las ventas de línea blanca.")
else:
    print(f"\n→ No se rechaza H0 (p = {p_valor_F:.4f} ≥ 0.05): modelo no significativo.")
```
